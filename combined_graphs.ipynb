```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

in_data = pd.read_csv('data/in_combined.csv')
out_data = pd.read_csv('data/combined.csv')

plt.plot(in_data['time'], in_data['pm10 std'])
plt.plot(out_data['time'], out_data['pm10 std'])
plt.plot(in_data['time'], in_data['pm25 std'])
plt.plot(out_data['time'], out_data['pm25 std'])
plt.plot(in_data['time'], in_data['pm100 std'])
plt.plot(out_data['time'], out_data['pm100 std'])
plt.plot(in_data['time'], in_data['Temperature'])
plt.plot(out_data['time'], out_data['Temperature'])
plt.plot(in_data['time'], in_data['Gas'])
plt.plot(out_data['time'], out_data['Gas'])
plt.plot(in_data['time'], in_data['Humidity'])
plt.plot(out_data['time'], out_data['Humidity'])
plt.plot(in_data['time'], in_data['Pressure'])
plt.plot(out_data['time'], out_data['Pressure'])
plt.plot(in_data['time'], in_data['Altitude'])
plt.plot(out_data['time'], out_data['Altitude'])

plt.title('Data of inside and outside vs Time')
plt.legend(loc='upper right', title="Legend", labels=['In pm10 std', 'Out pm10 std', 'In pm25 std', 'Out pm25 std', 'In pm100 std', 'Out pm100 std', 'In Temp', 'Out Temp', 'In Gas', 'Out Gas', 'In Humid', 'Out Humid', 'In Pressure', 'Out Pressure', 'In Alt', 'Out Alt'])
plt.xlabel('time')
plt.ylabel('data')
plt.show()

plt.hist(in_data['pm10 std'], bins = 4, label = 'Outside Data', color='blue', alpha = 0.5)
plt.hist(out_data['pm10 std'], bins = 4, label = 'Inside Data', color='red', alpha = 0.5)
plt.ylabel('Frequency')
plt.xlabel('pm10 std')
plt.title('Frequency of pm10 std Inside vs Outside')
plt.legend(loc = 'upper right')
plt.show()

plt.hist(in_data['pm25 std'], bins = 4, label = 'Outside Data', color='blue', alpha = 0.5)
plt.hist(out_data['pm25 std'], bins = 4, label = 'Inside Data', color='red', alpha = 0.5)
plt.ylabel('Frequency')
plt.xlabel('pm25 std')
plt.title('Frequency of pm25 std Inside vs Outside')
plt.legend(loc = 'upper right')
plt.show()

plt.hist(in_data['pm100 std'], bins = 4, label = 'Outside Data', color='blue', alpha = 0.5)
plt.hist(out_data['pm100 std'], bins = 4, label = 'Inside Data', color='red', alpha = 0.5)
plt.ylabel('Frequency')
plt.xlabel('pm100 std')
plt.title('Frequency of pm100 std Inside vs Outside')
plt.legend(loc = 'upper right')
plt.show()

plt.hist(in_data['Temperature'], bins = 4, label = 'Outside Data', color='blue', alpha = 0.5)
plt.hist(out_data['Temperature'], bins = 4, label = 'Inside Data', color='red', alpha = 0.5)
plt.ylabel('Frequency')
plt.xlabel('Temp')
plt.title('Frequency of Temperatures Inside vs Outside')
plt.legend(loc = 'upper right')
plt.show()

plt.hist(in_data['Gas'], bins = 4, label = 'Outside Data', color='blue', alpha = 0.5)
plt.hist(out_data['Gas'], bins = 4, label = 'Inside Data', color='red', alpha = 0.5)
plt.ylabel('Frequency')
plt.xlabel('Gas')
plt.title('Frequency of Gas Inside vs Outside')
plt.legend(loc = 'upper right')
plt.show()

plt.hist(in_data['Humidity'], bins = 4, label = 'Outside Data', color='blue', alpha = 0.5)
plt.hist(out_data['Humidity'], bins = 4, label = 'Inside Data', color='red', alpha = 0.5)
plt.ylabel('Frequency')
plt.xlabel('Humidity')
plt.title('Frequency of Humidity levels Inside vs Outside')
plt.legend(loc = 'upper right')
plt.show()

plt.hist(in_data['Pressure'], bins = 4, label = 'Outside Data', color='blue', alpha = 0.5)
plt.hist(out_data['Pressure'], bins = 4, label = 'Inside Data', color='red', alpha = 0.5)
plt.ylabel('Frequency')
plt.xlabel('Pressure')
plt.title('Frequency of Pressures Inside vs Outside')
plt.legend(loc = 'upper right')
plt.show()
```
