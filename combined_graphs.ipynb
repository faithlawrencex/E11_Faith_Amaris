```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

in_data = pd.read_csv('data/in_combined.csv')
out_data = pd.read_csv('data/combined.csv')

plt.plot(in_data['time'], in_data['par05'])
plt.plot(out_data['time'], out_data['par05'])
plt.title('Data of inside and outside vs Time')
plt.legend(loc='upper right', title="Legend", labels=['Inside Data', 'Outside Data'])
plt.xlabel('time')
plt.ylabel('particles 05um')
plt.show()

plt.hist(in_data['par05'], bins = 4, label = 'Inside Data', color='blue', alpha = 0.5)
plt.hist(out_data['par05'], bins = 4, label = 'Outside Data', color='red', alpha = 0.5)
plt.ylabel('Frequency')
plt.xlabel('particles 05um')
plt.title('Frequency of 05um particle Inside vs Outside')
plt.legend(loc = 'upper right')
plt.show()
```
