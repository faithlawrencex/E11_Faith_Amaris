```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

my_data = pd.read_csv('data/in_combined.csv')
other_data = pd.read_csv('data/other_group.csv')

plt.plot(my_data['time'], my_data['pm10 std'])
plt.plot(other_data['time'], other_data['pm10standard'])
plt.plot(my_data['time'], my_data['pm25 std'])
plt.plot(other_data['time'], other_data['pm25standard'])
plt.plot(my_data['time'], my_data['pm100 std'])
plt.plot(other_data['time'], other_data['pm100standard'])
plt.plot(my_data['time'], my_data['Temperature'])
plt.plot(other_data['time'], other_data['temperature'])
plt.plot(my_data['time'], my_data['Gas'])
plt.plot(other_data['time'], other_data['gas'])
plt.plot(my_data['time'], my_data['Humidity'])
plt.plot(other_data['time'], other_data['humidity'])
plt.plot(my_data['time'], my_data['Pressure'])
plt.plot(other_data['time'], other_data['pressure'])
plt.plot(my_data['time'], my_data['Altitude'])
plt.plot(other_data['time'], other_data['altitude'])

plt.title('Data of inside and outside vs Time')
plt.legend(loc='upper right', title="Legend", labels=['In pm10 std', 'Out pm10 std', 'In pm25 std', 'Out pm25 std', 'In pm100 std', 'Out pm100 std', 'In Temp', 'Out Temp', 'In Gas', 'Out Gas', 'In Humid', 'Out Humid', 'In Pressure', 'Out Pressure', 'In Alt', 'Out Alt'])
plt.xlabel('time')
plt.ylabel('data')
plt.show()

plt.hist(my_data['pm10 std'], bins = 4, label = 'Outside Data', color='blue', alpha = 0.5)
plt.hist(other_data['pm10standard'], bins = 4, label = 'Inside Data', color='red', alpha = 0.5)
plt.ylabel('Frequency')
plt.xlabel('pm10 std')
plt.title('Frequency of pm10 std Inside vs Outside')
plt.legend(loc = 'upper right')
plt.show()

plt.hist(my_data['pm25 std'], bins = 4, label = 'Outside Data', color='blue', alpha = 0.5)
plt.hist(other_data['pm25standard'], bins = 4, label = 'Inside Data', color='red', alpha = 0.5)
plt.ylabel('Frequency')
plt.xlabel('pm25 std')
plt.title('Frequency of pm25 std Inside vs Outside')
plt.legend(loc = 'upper right')
plt.show()

plt.hist(my_data['pm100 std'], bins = 4, label = 'Outside Data', color='blue', alpha = 0.5)
plt.hist(other_data['pm100standard'], bins = 4, label = 'Inside Data', color='red', alpha = 0.5)
plt.ylabel('Frequency')
plt.xlabel('pm100 std')
plt.title('Frequency of pm100 std Inside vs Outside')
plt.legend(loc = 'upper right')
plt.show()

plt.hist(my_data['Temperature'], bins = 4, label = 'Outside Data', color='blue', alpha = 0.5)
plt.hist(other_data['temperature'], bins = 4, label = 'Inside Data', color='red', alpha = 0.5)
plt.ylabel('Frequency')
plt.xlabel('Temp')
plt.title('Frequency of Temperatures Inside vs Outside')
plt.legend(loc = 'upper right')
plt.show()

plt.hist(my_data['Gas'], bins = 4, label = 'Outside Data', color='blue', alpha = 0.5)
plt.hist(other_data['gas'], bins = 4, label = 'Inside Data', color='red', alpha = 0.5)
plt.ylabel('Frequency')
plt.xlabel('Gas')
plt.title('Frequency of Gas Inside vs Outside')
plt.legend(loc = 'upper right')
plt.show()

plt.hist(my_data['Humidity'], bins = 4, label = 'Outside Data', color='blue', alpha = 0.5)
plt.hist(other_data['humidity'], bins = 4, label = 'Inside Data', color='red', alpha = 0.5)
plt.ylabel('Frequency')
plt.xlabel('Humidity')
plt.title('Frequency of Humidity levels Inside vs Outside')
plt.legend(loc = 'upper right')
plt.show()

plt.hist(my_data['Pressure'], bins = 4, label = 'Outside Data', color='blue', alpha = 0.5)
plt.hist(other_data['pressure'], bins = 4, label = 'Inside Data', color='red', alpha = 0.5)
plt.ylabel('Frequency')
plt.xlabel('Pressure')
plt.title('Frequency of Pressures Inside vs Outside')
plt.legend(loc = 'upper right')
plt.show()
```
