```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

#plotting temp vs humidity for outside data we got with 60 sec delayed start
data = pd.read_csv('data/combined.csv')

plt.plot(data['Temperature'], data['Humidity'])
plt.title('Temperature vs Humidity')
plt.xlabel('Temperature')
plt.ylabel('Humidity')
plt.show()

plt.scatter(data['Temperature'], data['Humidity'], alpha=0.5)
plt.title('Temperature vs Humidity')
plt.xlabel('Temperature')
plt.ylabel('Humidity')
plt.show()
```

The hidden variable that is important when making plots like this is that time they are taken has to line up. If the temp associated with a humidity value was taken at different times, it changes the correlation and makes it wrong. This plot seems to indicate a negagive linear relationship between temperature and humidity.

```python
Temp = np.array(data['Temperature'])
Hum = np.array(data['Humidity'])

correlation_matrix = np.corrcoef(Temp, Hum)
r = correlation_matrix[0, 1]
r_squared = r**2

print(f"Pearson r: {r:.4f}")
print(f"R-squared: {r_squared:.4f}")
```

```python
plt.plot(data['Temperature'], data['Pressure'])
plt.title('Temperature vs Pressure')
plt.xlabel('Temperature')
plt.ylabel('Pressure')
plt.show()

Temp = np.array(data['Temperature'])
Pres = np.array(data['Pressure'])

correlation_matrix = np.corrcoef(Temp, Pres)
r = correlation_matrix[0, 1]
r_squared = r**2

print(f"Pearson r: {r:.4f}")
print(f"R-squared: {r_squared:.4f}")
```

It is similar to temp v humidity as is is pretty linear with similar correlation values, but it has a positive correlation instead of negative

```python
plt.plot(data['Temperature'], data['pm25 std'])
plt.title('Temperature vs pm25 std')
plt.xlabel('Temperature')
plt.ylabel('pm25 std')
plt.show()

Temp = np.array(data['Temperature'])
pm25 = np.array(data['pm25 std'])

correlation_matrix = np.corrcoef(Temp, pm25)
r = correlation_matrix[0, 1]
r_squared = r**2

print(f"Pearson r: {r:.4f}")
print(f"R-squared: {r_squared:.4f}")
```
