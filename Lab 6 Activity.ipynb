```python
import numpy as np
import pandas as pd

my_data = pd.read_csv('combined.csv')

numeric_data = my_data.select_dtypes(include=[np.number])

n = len(numeric_data)

means = numeric_data.mean()
stds = numeric_data.std()

mean_uncertainty = stds / np.sqrt(n)

print("Means:")
print(means)

print("\nStandard Deviations:")
print(stds)

print("\nUncertainty on the Mean:")
print(mean_uncertainty)
```

```python
data = pd.read_csv('combined.csv')

half = len(data) // 2

inside = data.iloc[:half]
outside = data.iloc[half:]

inside_num = inside.select_dtypes(include=[np.number])
outside_num = outside.select_dtypes(include=[np.number])

mean_inside = inside_num.mean()
mean_outside = outside_num.mean()

std_inside = inside_num.std()
std_outside = outside_num.std()

n_inside = len(inside_num)
n_outside = len(outside_num)

unc_inside = std_inside / np.sqrt(n_inside)
unc_outside = std_outside / np.sqrt(n_outside)

sigma_sep = abs(mean_inside - mean_outside) / np.sqrt(unc_inside**2 + unc_outside**2)

print("Sigma Separation:")
print(sigma_sep)
```

3b. Most variables were separated by less than 2σ, indicating reasonable agreement between indoor and outdoor measurements. However, some variables showed larger separations, suggesting statistically significant environmental differences.

3c. Our distributions were not completely Gaussian which suggests that the sample size may not have been large enough for the Central Limit Theorem to fully apply. Increasing the number of measurements would help the distribution of the sample mean approach a normal distribution.

```python
import numpy as np
import pandas as pd

my_data = pd.read_csv("combined.csv")
other_data = pd.read_csv("other_group.csv")

half_my = len(my_data) // 2
half_other = len(other_data) // 2

my_inside = my_data.iloc[:half_my]
my_outside = my_data.iloc[half_my:]

other_inside = other_data.iloc[:half_other]
other_outside = other_data.iloc[half_other:]

my_inside = my_inside.select_dtypes(include=[np.number])
my_outside = my_outside.select_dtypes(include=[np.number])
other_inside = other_inside.select_dtypes(include=[np.number])
other_outside = other_outside.select_dtypes(include=[np.number])

mean_my_inside = my_inside.mean()
mean_other_inside = other_inside.mean()

std_my_inside = my_inside.std()
std_other_inside = other_inside.std()

n_my_inside = len(my_inside)
n_other_inside = len(other_inside)

unc_my_inside = std_my_inside / np.sqrt(n_my_inside)
unc_other_inside = std_other_inside / np.sqrt(n_other_inside)

sigma_inside = abs(mean_my_inside - mean_other_inside) / np.sqrt(
    unc_my_inside**2 + unc_other_inside**2
)

mean_my_outside = my_outside.mean()
mean_other_outside = other_outside.mean()

std_my_outside = my_outside.std()
std_other_outside = other_outside.std()

n_my_outside = len(my_outside)
n_other_outside = len(other_outside)

unc_my_outside = std_my_outside / np.sqrt(n_my_outside)
unc_other_outside = std_other_outside / np.sqrt(n_other_outside)

sigma_outside = abs(mean_my_outside - mean_other_outside) / np.sqrt(
    unc_my_outside**2 + unc_other_outside**2
)

print("Inside Means (My Group):")
print(mean_my_inside)

print("\nInside Means (Other Group):")
print(mean_other_inside)

print("\nInside Sigma Separation:")
print(sigma_inside)

print("\nOutside Means (My Group):")
print(mean_my_outside)

print("\nOutside Means (Other Group):")
print(mean_other_outside)

print("\nOutside Sigma Separation:")
print(sigma_outside)
```

The measurements from our sensors and the other group’s sensors were compared using the difference in means relative to the combined uncertainty. Most variables differed by less than 2σ, indicating that the sensors produced statistically consistent measurements. This suggests that the sensors have good precision and are reasonably well calibrated.
