```python
import numpy as np
import pandas as pd

my_data = pd.read_csv('combined.csv')

numeric_data = my_data.select_dtypes(include=[np.number])

n = len(numeric_data)

means = numeric_data.mean()
stds = numeric_data.std()

mean_uncertainty = stds / np.sqrt(n)

print("Means:")
print(means)

print("\nStandard Deviations:")
print(stds)

print("\nUncertainty on the Mean:")
print(mean_uncertainty)
```

```python
data = pd.read_csv('combined.csv')

half = len(data) // 2

inside = data.iloc[:half]
outside = data.iloc[half:]

inside_num = inside.select_dtypes(include=[np.number])
outside_num = outside.select_dtypes(include=[np.number])

mean_inside = inside_num.mean()
mean_outside = outside_num.mean()

std_inside = inside_num.std()
std_outside = outside_num.std()

n_inside = len(inside_num)
n_outside = len(outside_num)

unc_inside = std_inside / np.sqrt(n_inside)
unc_outside = std_outside / np.sqrt(n_outside)

sigma_sep = abs(mean_inside - mean_outside) / np.sqrt(unc_inside**2 + unc_outside**2)

print("Sigma Separation:")
print(sigma_sep)
```

3b. Most variables were separated by less than 2σ, indicating reasonable agreement between indoor and outdoor measurements. However, some variables showed larger separations, suggesting statistically significant environmental differences.

3c. Our distributions were not completely Gaussian which suggests that the sample size may not have been large enough for the Central Limit Theorem to fully apply. Increasing the number of measurements would help the distribution of the sample mean approach a normal distribution.
